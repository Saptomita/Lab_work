Problem 2: The Water Jug Problem (Solved using BFS)
Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque


# Capacity of the jugs
CAPACITY_A = 4
CAPACITY_B = 3


def water_jug_bfs():

    # Starting state
    # A = 0 litres, B = 0 litres
    start = (0, 0)

    # Queue contains:
    # (current_state, list_of_actions)
    queue = deque()
    queue.append((start, []))

    # Set to store already visited states
    visited = set()
    visited.add(start)

    while queue:

        # Remove the first state from the queue
        (a, b), path = queue.popleft()

        # Check whether we have exactly 2 litres
        if a == 2 or b == 2:
            return path, (a, b)

        # List of possible operations
        next_states = []

        # ------------------------------------------------
        # 1. Fill Jug A
        # ------------------------------------------------
        next_states.append(
            ((CAPACITY_A, b), "Fill Jug A")
        )

        # ------------------------------------------------
        # 2. Fill Jug B
        # ------------------------------------------------
        next_states.append(
            ((a, CAPACITY_B), "Fill Jug B")
        )

        # ------------------------------------------------
        # 3. Empty Jug A
        # ------------------------------------------------
        next_states.append(
            ((0, b), "Empty Jug A")
        )

        # ------------------------------------------------
        # 4. Empty Jug B
        # ------------------------------------------------
        next_states.append(
            ((a, 0), "Empty Jug B")
        )

        # ------------------------------------------------
        # 5. Pour Jug A into Jug B
        # ------------------------------------------------

        # Amount that can be poured from A to B
        transfer = min(a, CAPACITY_B - b)

        new_a = a - transfer
        new_b = b + transfer

        next_states.append(
            ((new_a, new_b), "Pour Jug A -> Jug B")
        )

        # ------------------------------------------------
        # 6. Pour Jug B into Jug A
        # ------------------------------------------------

        # Amount that can be poured from B to A
        transfer = min(b, CAPACITY_A - a)

        new_a = a + transfer
        new_b = b - transfer

        next_states.append(
            ((new_a, new_b), "Pour Jug B -> Jug A")
        )

        # ------------------------------------------------
        # Add new states to the BFS queue
        # ------------------------------------------------

        for state, action in next_states:

            # Only visit a state once
            if state not in visited:

                visited.add(state)

                # Add the new state and action to the queue
                queue.append(
                    (state, path + [action])
                )


# --------------------------------------------------------
# Run the BFS algorithm
# --------------------------------------------------------

actions, final_state = water_jug_bfs()


# --------------------------------------------------------
# Display the solution
# --------------------------------------------------------

print("Water Jug Problem")
print("=================")

print("\nJug A capacity = 4 litres")
print("Jug B capacity = 3 litres")

print("\nMinimum sequence of steps:\n")

current_a = 0
current_b = 0

print(f"Start: ({current_a}, {current_b})")
print("       A = 0 L, B = 0 L")

for step, action in enumerate(actions, start=1):

    # Find the resulting state by performing the action
    if action == "Fill Jug A":
        current_a = 4

    elif action == "Fill Jug B":
        current_b = 3

    elif action == "Empty Jug A":
        current_a = 0

    elif action == "Empty Jug B":
        current_b = 0

    elif action == "Pour Jug A -> Jug B":
        transfer = min(current_a, 3 - current_b)
        current_a -= transfer
        current_b += transfer

    elif action == "Pour Jug B -> Jug A":
        transfer = min(current_b, 4 - current_a)
        current_a += transfer
        current_b -= transfer

    print(f"\nStep {step}: {action}")
    print(f"       State: ({current_a}, {current_b})")
    print(f"       Jug A = {current_a} L")
    print(f"       Jug B = {current_b} L")


print("\n=================")
print("Solution found!")
print(f"Final state: {final_state}")
print(f"Minimum number of steps: {len(actions)}")
```
